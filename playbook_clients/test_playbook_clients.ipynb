# Clients: 12 минимальных smoke-тестов

Ноутбук демонстрирует ленивую фабрику `zemi.playbook.clients.Clients`: создаёт каждый из 12 поддерживаемых клиентов, проверяет кэширование свойства и выводит фактический тип объекта. Сетевые запросы не выполняются, поэтому запущенный llama.cpp-сервер не требуется.

```python
import sys
from pathlib import Path

PROJECT_ROOT = next(
    directory
    for directory in (Path.cwd(), *Path.cwd().parents)
    if (directory / ".zemicomp").is_file() and (directory / "zemi").is_dir()
)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from zemi.playbook.clients import Clients
```

```python
clients = Clients(
    "http://127.0.0.1:8080/v1",
    model="test-model",
    api_key="llama.cpp",
    timeout=5.0,
)

assert clients.server_url == "http://127.0.0.1:8080"
assert clients.openai_url == "http://127.0.0.1:8080/v1"
assert len(clients.names) == 12

results = {}

def smoke_test(name):
    client = getattr(clients, name)
    assert client is getattr(clients, name), f"{name} не кэшируется"
    type_name = f"{type(client).__module__}.{type(client).__name__}"
    results[name] = type_name
    print(f"PASS {name:16} -> {type_name}")
    return client
```
