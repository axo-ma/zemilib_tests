# Clients: 12 минимальных smoke-тестов

Ноутбук демонстрирует ленивую фабрику `zemi.arsenal.clients.Clients`: создаёт каждый из 12 поддерживаемых клиентов, проверяет кэширование свойства и выводит фактический тип объекта. Сетевые запросы не выполняются, поэтому запущенный llama.cpp-сервер не требуется.

```python
from zemi.arsenal.clients import Clients
```

```python
clients = Clients(
    "http://127.0.0.1:8080/v1",
    model="test-model",
    api_key="llama.cpp",
    timeout=5.0,
)

assert clients.server_url == "http://127.0.0.1:8080"
assert clients.openai_url == "http://127.0.0.1:8080/v1"
assert len(clients.names) == 12

results = {}

def smoke_test(name):
    client = getattr(clients, name)
    assert client is getattr(clients, name), f"{name} не кэшируется"
    type_name = f"{type(client).__module__}.{type(client).__name__}"
    results[name] = type_name
    print(f"PASS {name:16} -> {type_name}")
    return client
```
